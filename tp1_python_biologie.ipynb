```python
import numpy as np

heures = np.array([0, 1, 2, 3, 4, 5])
cellules = np.array([100, 150, 210, 300, 430, 600])

print("Heures :", heures)
print("Cellules :", cellules)
```

```python
moyenne = np.mean(cellules)

print("La moyenne des cellules est :", moyenne)
```

```python
import matplotlib.pyplot as plt

plt.plot(heures, cellules, marker='o')
plt.xlabel("Heures")
plt.ylabel("Nombre de cellules")
plt.title("Evolution des cellules")
plt.grid()
plt.show()
```
